# Trajectory of the spectral parameter E under a Cattaneo constitutive law

**Notebook 01** — first result of the internship.

Cattaneo replaces the spectral parameter `p` by `p + tau*p^2`. In the modulated regime
`p = i*omega`, so `p_eff = -tau*omega^2 + i*omega` and the energy of the Schrödinger
analogy is `E = -p_eff = tau*omega^2 - i*omega`.

Writing `x = omega*tau` gives `E = omega*(x - i)`. The factor `omega` is real and positive,
so it does not change the argument: any positive real multiple of `(x - i)` traces the same
direction. The choice `x/(x+i)` gives an exact half-circle of radius 1/2 centred on 1/2 —
the Cole-Cole diagram of a single-relaxation process.

**Control points to verify:** −89.43° at `x = 0.01`, exactly −45° at `x = 1`, −0.57° at `x = 100`.


```python
import numpy as np
import matplotlib.pyplot as plt

# ---------------------------------------------------------------
# 1. Le parametre sans dimension
# ---------------------------------------------------------------
x = np.logspace(-2, 2, 1001)          # omega*tau de 0.01 a 100

# ---------------------------------------------------------------
# 2. Les deux ecritures de E
# ---------------------------------------------------------------
tau = 1.0                             # tau = 1 : seul le produit omega*tau compte
omega = x / tau

E_deriv = tau * omega**2 - 1j * omega  # l'energie telle qu'elle sort de la derivation
E_cole = x / (x + 1j)                  # la meme direction, module normalise en demi-cercle

arg_deriv = np.angle(E_deriv, deg=True)
arg_cole = np.angle(E_cole, deg=True)

# ---------------------------------------------------------------
# 3. Verifications numeriques
# ---------------------------------------------------------------
print("Points de controle")
print(f"{'omega*tau':>12} {'arg(E) derive':>16} {'arg(E) Cole':>14}")
for cible in (0.01, 1.0, 100.0):
    k = int(np.argmin(np.abs(x - cible)))
    print(f"{x[k]:12.4f} {arg_deriv[k]:16.4f} {arg_cole[k]:14.4f}")

ecart = np.max(np.abs(arg_deriv - arg_cole))
print(f"\nEcart maximal entre les deux arguments : {ecart:.2e} degre")

rayon = np.abs(E_cole - 0.5)
print(f"Ecart maximal au demi-cercle de rayon 1/2 : {np.max(np.abs(rayon - 0.5)):.2e}")

# ---------------------------------------------------------------
# 4. Figure
# ---------------------------------------------------------------
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11.5, 4.8))

# --- Panneau A : le plan complexe -------------------------------
ax1.plot(E_cole.real, E_cole.imag, lw=2, color="#1f6f8b")
ax1.axhline(0, lw=0.8, color="0.4")
ax1.axvline(0, lw=0.8, color="0.4")

for cible, dx, dy in ((0.01, 10, 10), (1.0, 10, -18), (100.0, -95, 10)):
    k = int(np.argmin(np.abs(x - cible)))
    ax1.plot(E_cole.real[k], E_cole.imag[k], "o", ms=7, color="#b03030")
    ax1.annotate(f"$\\omega\\tau$ = {cible:g}\n{arg_cole[k]:.2f}$^\\circ$",
                 (E_cole.real[k], E_cole.imag[k]),
                 textcoords="offset points", xytext=(dx, dy), fontsize=9)

ax1.set_xlabel("partie réelle (module normalisé)")
ax1.set_ylabel("partie imaginaire (module normalisé)")
ax1.set_title("Direction de E dans le plan complexe")
ax1.set_aspect("equal")
ax1.set_xlim(-0.15, 1.25)
ax1.set_ylim(-0.65, 0.25)
ax1.grid(alpha=0.3)

# --- Panneau B : la transition ----------------------------------
ax2.semilogx(x, arg_deriv, lw=2, color="#1f6f8b")
ax2.axhline(-45, ls="--", lw=1, color="#b03030")
ax2.axvline(1, ls="--", lw=1, color="#b03030")
ax2.annotate("$-45^\\circ$ à $\\omega\\tau = 1$", (1, -45),
             textcoords="offset points", xytext=(10, 12),
             fontsize=9, color="#b03030")
ax2.set_xlabel("$\\omega\\tau$")
ax2.set_ylabel("arg(E)  [degrés]")
ax2.set_title("Transition : de l'axe imaginaire vers l'axe réel")
ax2.set_ylim(-95, 5)
ax2.grid(alpha=0.3, which="both")

plt.tight_layout()
plt.savefig("../figures/01_cattaneo_energy_trajectory.png", dpi=150)
plt.show()
```
